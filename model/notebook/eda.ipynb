```python
drug_cols = ["troglitazone", "tolbutamide", "acetohexamide", "examide",
             "citoglipton", "insulin", "glyburide-metformin",
             "glipizide-metformin", "glimepiride-pioglitazone",
             "metformin-rosiglitazone"]
present = [c for c in drug_cols if c in df.columns]
print("Missing from df:", set(drug_cols) - set(present))
print(df[present].isna().sum())
for c in present:
    print(c, df[c].value_counts(dropna=False).to_dict())
```

```python
for c in ["change", "diabetesMed"]:
    print(c, df[c].value_counts(dropna=False).to_dict())
print(df[["change", "diabetesMed"]].isna().sum())
```
